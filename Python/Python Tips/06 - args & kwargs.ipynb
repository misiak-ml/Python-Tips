<b>Deň - 06 - *args a **kwargs</b>

Špeciálne syntaxy ``*args`` a ``**kwargs`` sa používajú na odovzdávanie viariabilného počtu argumentov. 

V prípade použitia ``*args`` <b>positional argument</b> sa vo funkcii sa všetky argumenty zhromaždia do tuple-u a počet neoznačených argumentov. Ich počet je ľubovolný. 


Hlavný rozdiel medzi ``*args`` a ``*kwargs`` je, že sa argumenty args su odovzdané bez názvu, a teda len čisto ako hodnty.


```python
def objednaj_jedlo(*jedla):
    print("Objednané jedlá:", jedla)
    
    return jedla

a = objednaj_jedlo("pizza", "burger", "sushi")
```

V prípade použitia ``**kwargs`` sa vo funkcii sa všetky argumenty zhromaždia do slovníka, kde kĺuče sú názvy argumentov a values sú ich hodnoty. Počet argumentov, ktoré idú do ``**kwargs`` je rovnako ľubovolný.

```python
def objednaj_jedlo_s_detailmi(**detaily):
    print("Detaily objednávky:", detaily)

objednaj_jedlo_s_detailmi(jedlo="šunková pizza", veľkosť="maxi", dressing="cesnak")
```

```python
#Bez pouzitia *args 
def scitaj_dve_cisla(a, b): #-> funkcia sumarizuje len cez dve premenne
    return sum([a, b])

print(scitaj_dve_cisla(3, 5))  
# print(scitaj_dve_cisla(5)) -> vrati chybu
```

```python
def scitaj(*args): #-> funkcia prijima variabilny pocet vstupov a 
    return sum(args)

print(scitaj(1, 2, 9))
print(scitaj(0))
```

```python
#Bez pouzitia **kwargs 
def vypis_info(meno, vek):
    print(f"meno: {meno}\nvek: {vek}")

vypis_info("Michal", 30)  
```

```python
def vypis_info(**kwargs):
    for k, v in kwargs.items():
        print(f"{k}: {v}")

vypis_info(meno="Jano", vek=30, adresa="Lucky")
```

```python
def popis_osoby(meno, *args, **kwargs):
    print("Meno:", meno)
    print("Dalšie hodnoty:", args)
    print("Extra info:", kwargs)
    
    return f"Meno: {meno} Dalšie hodnoty: {args} extra info:{kwargs}"

o = popis_osoby("Jano", 33, 194, povolanie="programator", mesto="Bratislava")
```

Ukážka komplexnejšieho príkladu využitia s využitím <b>dekorátora</b> - tomu sa budem venovať v ďalšej ukážke

```python
def spracuj_objednavku(stol, *jedla, **detaily):
    """
    Funkcia na spracovanie objednávky v reštaurácii.
    
    Parametre:
    - stol: Číslo stola (povinný argument)
    - *jedla: Variabilný počet jedál (napr. "pizza", "burger", "pasta")
    - **detaily: Voliteľné detaily objednávky (napr. cas_dorucenia, tip, poznamka)
    """
    
    print(f"\nObjednávka pre stôl č. {stol}:")
    
    # Spracovanie jedál (*args)
    if jedla:
        print("Objednané jedlá:")
        for idx, jedlo in enumerate(jedla, 1):
            print(f"  {idx}. {jedlo}")
    else:
        print("Neboli objednané žiadne jedlá.")
    
    # Spracovanie voliteľných detailov (**kwargs)
    print("Detaily objednávky:")
    if detaily:
        for klic, hodnota in detaily.items():
            print(f"  {klic.replace('_', ' ').capitalize()}: {hodnota}")
    else:
        print("  Žiadne dodatočné detaily.")
    
    # Výpočet ceny (príklad: každé jedlo stojí 10 € + prípadný tip)
    cena_jedla = len(jedla) * 10
    tip = detaily.get("tip", 0)  # Ak je tip v detaily, použije sa, inak 0
    celkova_cena = cena_jedla + tip
    
    print(f"Celková cena: {celkova_cena} € (jedlo: {cena_jedla} € + tip: {tip} €)")

# Dekorátor na logovanie objednávok
def log_objednavku(func):
    def wrapper(*args, **kwargs):
        print("=== Nová objednávka ===")
        vysledok = func(*args, **kwargs)
        print("=====================")
        return vysledok
    return wrapper
```

```python
@log_objednavku
def vytvor_objednavku(stol, *jedla, **detaily):
    spracuj_objednavku(stol, *jedla, **detaily)


vytvor_objednavku(1, "pizza", "pasta", cas_dorucenia="18:00", tip=5)
    
vytvor_objednavku(2, poznamka="Len nápoje")
    
vytvor_objednavku(
        3,
        "burger", "steak", "salát",
        cas_dorucenia="19:30",
        tip=10,
        specialna_poziadavka="Bez cibule",
        prilahly_stol=4
    )
```
